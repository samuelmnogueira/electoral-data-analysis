```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Year": [2004, 2009, 2014, 2019, 2024],

    # Portugal
    "PT_Turnout": [63, 59, 57, 49, 59],
    "PT_Trust": [45, 38, 32, 28, 40],

    # EU média
    "EU_Turnout": [70, 72, 68, 66, 70],
    "EU_Trust": [55, 52, 48, 45, 50]
})

df
```

```python
plt.plot(df["Year"], df["PT_Turnout"], marker="o", label="Portugal")
plt.plot(df["Year"], df["EU_Turnout"], marker="o", label="EU Average")

plt.title("Voter Turnout: Portugal vs EU")
plt.xlabel("Year")
plt.ylabel("Turnout (%)")
plt.legend()

plt.show()
```

```python
plt.plot(df["Year"], df["PT_Trust"], marker="o", label="Portugal")
plt.plot(df["Year"], df["EU_Trust"], marker="o", label="EU Average")

plt.title("Political Trust: Portugal vs EU")
plt.xlabel("Year")
plt.ylabel("Trust Index (0-100)")
plt.legend()

plt.show()
```

```python
plt.scatter(df["PT_Trust"], df["PT_Turnout"], label="Portugal")
plt.scatter(df["EU_Trust"], df["EU_Turnout"], label="EU")

plt.title("Trust vs Electoral Participation")
plt.xlabel("Political Trust")
plt.ylabel("Turnout")
plt.legend()

plt.show()
```

## Political Analysis

This project compares Portugal and the EU average in terms of voter turnout and political trust.

The results suggest:
- Portugal shows a stronger decline in political trust over time.
- Lower trust is associated with lower voter turnout.
- The EU average remains more stable in both indicators.

These patterns suggest a potential relationship between institutional trust and democratic participation.
