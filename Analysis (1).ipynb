```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "country": ["Portugal", "Spain", "France", "Germany", "Italy"],
    "participation": [55, 68, 65, 76, 60],
    "unemployment": [6.5, 12.3, 7.1, 5.2, 9.8],
    "income": [22000, 24000, 26000, 30000, 23000]
}

df = pd.DataFrame(data)
df
```

```python
df.head()
```

```python
df.describe()
```

```python
df["participation"].mean()
```

```python
df.corr(numeric_only=True)
```

```python
plt.scatter(df["unemployment"], df["participation"])
plt.xlabel("Unemployment Rate")
plt.ylabel("Electoral Participation")
plt.title("Unemployment vs Electoral Participation")
plt.show()
```

```python
plt.bar(df["country"], df["participation"])
plt.title("Electoral Participation by Country")
plt.xticks(rotation=45)
plt.show()
```

## Analysis

This project explores the relationship between socioeconomic factors and electoral participation.

The data suggests that differences in unemployment and income levels may be associated with variations in voter turnout across countries.

## Conclusion

Economic factors appear to have an influence on electoral participation. Further analysis with larger datasets would be useful to confirm these patterns.
