```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("data_portugal.xlsx", sheet_name="Folha1")

df["Year"] = df["Year"].astype(int)
df["Unemployment"] = pd.to_numeric(df["Unemployment"])
df["Turnout"] = pd.to_numeric(df["Turnout"])

print(df)

plt.figure(figsize=(10,5))

plt.plot(df["Year"], df["Unemployment"], marker="o", label="Unemployment (%)")
plt.plot(df["Year"], df["Turnout"], marker="o", label="Voter Turnout (%)")

plt.title("Unemployment vs Voter Turnout in Portugal")
plt.xlabel("Year")
plt.ylabel("Percentage (%)")
plt.legend()

plt.xticks(df["Year"])

plt.show()

plt.figure(figsize=(7,5))

plt.scatter(df["Unemployment"], df["Turnout"])

plt.title("Unemployment vs Voter Turnout")
plt.xlabel("Unemployment Rate (%)")
plt.ylabel("Turnout (%)")

plt.show()

correlation = df[["Unemployment", "Turnout"]].corr()
print("\nCorrelation:\n", correlation)
```

## Analysis

This project examines the relationship between unemployment and voter turnout in Portugal using structured data inspired by PORDATA.

The time series analysis shows fluctuations in both unemployment and electoral participation over time.

The scatter plot suggests a potential negative relationship between unemployment and voter turnout, meaning higher unemployment may be associated with lower political participation.
